# Practical 1 - Revisiting Pandas and Numpy

```python
import numpy as np
import pandas as pd
```

Step 1: Take raw data and convert it into DataFrames

```python
raw_data = {
    'Employee_ID': [101, 102, 103, 104],
    'Department': ['HR', 'IT', 'IT', 'Finance'],
    'Salary': [50000, 75000, 80000, 65000],
    'Bonus': [5000, 7000, 8000, 6000]
}
df = pd.DataFrame(raw_data)

print("Created DataFrame:")
print(df)
```

Step 2: Using NumPy to find shape, size, and dimension

```python
arr = df[['Salary', 'Bonus']].to_numpy()

print("NumPy Array:")
print(arr)

print(arr.shape)
print(arr.size)
print(arr.ndim)
```

Step 3: Matrix Operations (Addition, Subtraction, Multiplication, Inverse, Transpose, Product)

```python
A = np.array([[4, 7], 
              [2, 6]])

B = np.array([[1, 3], 
              [2, 5]])

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)

add_res = np.add(A, B)
print("\nAddition : ", add_res)

sub_res = np.subtract(A, B)
print("Subtraction : ", sub_res)

elem_mult = np.multiply(A, B)
print("Multiplication : ", elem_mult)

transpose_A = np.transpose(A)
print("Transpose : ", transpose_A)

inverse_A = np.linalg.inv(A)
print("Inverse : ", inverse_A)
```
